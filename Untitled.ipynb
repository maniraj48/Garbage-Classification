```python
import tensorflow as tf
import numpy as np

# --- 1. Define Parameters ---
# --- IMPORTANT: UPDATE THIS PATH IF NEEDED ---
dataset_path = './Garbage classification/Garbage classification'

# Image and batching parameters that our model will use
IMG_HEIGHT = 224
IMG_WIDTH = 224
BATCH_SIZE = 32  # Number of images to process in one go


# --- 2. Create the Training and Validation Datasets ---
print("Creating datasets...")

# Create the training dataset (80% of the data)
# TensorFlow's utility function automatically infers class names from the
# directory structure and labels the data accordingly.
train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,  # Reserve 20% of the data for validation
    subset="training",
    seed=123,              # A seed for shuffling ensures reproducibility
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE
)

# Create the validation dataset (the remaining 20% of the data)
val_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2, # This split percentage must be the same as above
    subset="validation",
    seed=123,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE
)


# --- 3. Get the Class Names ---
class_names = train_ds.class_names
print(f"\nFound the following class names: {class_names}")
num_classes = len(class_names)
print(f"Number of classes: {num_classes}")


# --- 4. Create a Separate Test Set ---
# The `val_ds` we created contains all 20% of the reserved data.
# We will split it in half, using one half for validation during training,
# and the other half as a final, unseen test set.
val_batches = tf.data.experimental.cardinality(val_ds)
test_ds = val_ds.take(val_batches // 2)
val_ds = val_ds.skip(val_batches // 2)

print(f'\nNumber of training batches: {tf.data.experimental.cardinality(train_ds)}')
print(f'Number of validation batches: {tf.data.experimental.cardinality(val_ds)}')
print(f'Number of test batches: {tf.data.experimental.cardinality(test_ds)}')


# --- 5. Inspect a Batch of Data ---
# Let's see what a single batch of data looks like. A batch contains
# BATCH_SIZE images and their corresponding labels.
print("\n--- Inspecting a batch from the training data ---")
for images, labels in train_ds.take(1):
    print("Images batch shape:", images.shape)
    print("Labels batch shape:", labels.shape)
    print("\nFirst 5 labels in the batch:")
    for i in range(5):
        # The label is a number (0-5). We can use it as an index to get the class name.
        print(f"- Image {i+1} has label: {labels[i].numpy()} which corresponds to class: '{class_names[labels[i]]}'")


# --- 6. Configure Datasets for Performance ---
# These are standard optimization steps for a TensorFlow data pipeline.
# .cache() keeps images in memory after they're loaded, saving time.
# .prefetch() overlaps data preprocessing and model training.
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.cache().prefetch(buffer_size=AUTOTUNE)

print("\nDatasets have been successfully created, split, and configured for performance.")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# =============================================================================
# NOTE: Ensure you have already run the "Stage 2" data preprocessing code
# to create the train_ds, val_ds, and test_ds variables.
# =============================================================================


# --- 1. Define a Data Augmentation Layer ---
# This block of layers will be applied to each training image on the fly.
print("\n--- Defining data augmentation layers ---")
data_augmentation = keras.Sequential(
    [
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.1),
        layers.RandomZoom(0.1),
    ],
    name="data_augmentation",
)


# --- 2. Build the NEW CNN Model with Augmentation ---
print("--- Building the CNN model with data augmentation ---")
model = keras.Sequential([
    # Input Shape
    layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 3)),

    # Apply data augmentation ONLY during training, not during evaluation.
    # Keras handles this automatically.
    data_augmentation,

    # Rescale pixel values from [0, 255] to [0, 1]
    layers.Rescaling(1./255),

    # The rest of the model is the same as before
    layers.Conv2D(16, (3, 3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])


# --- 3. Compile the Model ---
print("--- Compiling the model ---")
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# --- 4. Print a Summary of the Model ---
print("Model Summary:")
model.summary()


# --- 5. Train the Model ---
epochs = 15 # Let's use the same number of epochs to compare fairly
print(f"\n--- Starting model training for {epochs} epochs ---")

history = model.fit(
  train_ds,
  validation_data=val_ds,
  epochs=epochs
)

print("\n--- Model training finished ---")


# =============================================================================
# NOW, RUN THE STAGE 4 EVALUATION CODE AGAIN TO SEE THE NEW RESULTS
# The visualization code from the previous step can be run without changes.
# =============================================================================
```

```python
import matplotlib.pyplot as plt

# --- Make sure the 'history' and 'test_ds' variables from the previous step are available ---

# --- 1. Visualize Training and Validation Metrics ---
print("\n--- Visualizing training history ---")

# Extracting the history from our training session
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(len(acc))

# Create a figure with two subplots
plt.figure(figsize=(12, 6))

# Subplot 1: Training and Validation Accuracy
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')

# Subplot 2: Training and Validation Loss
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')

# Show the plots
plt.suptitle('Model Training History', fontsize=16)
plt.show()


# --- 2. Evaluate the Model on the Test Set ---
print("\n--- Evaluating model on the unseen test set ---")

# The model.evaluate function returns the loss and the metrics
# we specified during compilation (in our case, just 'accuracy').
loss, accuracy = model.evaluate(test_ds)

# Print the final accuracy
print(f'\nTest Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy*100:.2f}%')
```
